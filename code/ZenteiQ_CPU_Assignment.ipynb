```python
# Cell 1: Complete working script with learnable patterns
import jax
import jax.numpy as jnp
from jax import random
import numpy as np
import time
import os

print("✅ JAX version:", jax.__version__)
print("✅ Device:", jax.devices()[0])

# Create synthetic dataset with PATTERNS (not random)
def create_pattern_data(batch_size=8, seq_len=32, vocab_size=100):
    """Create data with repeating patterns for the model to learn"""
    # Create patterns: repeat numbers 1,2,3,4,5,...
    pattern = jnp.tile(jnp.arange(1, 11), (seq_len // 10 + 1))[:seq_len] % vocab_size

    # Add some noise but keep the pattern
    noise = random.randint(random.PRNGKey(0), (seq_len,), 0, 10)
    pattern = (pattern + noise) % vocab_size

    # Tile for batch
    data = jnp.tile(pattern[None, :], (batch_size, 1))
    return data

# Model that can actually learn
class SimpleModel:
    def __init__(self, vocab_size=100, embed_dim=32):
        self.vocab_size = vocab_size
        self.embed_dim = embed_dim

    def init_params(self, key):
        keys = random.split(key, 2)
        embed = random.normal(keys[0], (self.vocab_size, self.embed_dim)) * 0.1
        dense = random.normal(keys[1], (self.embed_dim, self.vocab_size)) * 0.1
        return {'embed': embed, 'dense': dense}

    def forward(self, params, x):
        embed = params['embed'][x]  # (batch, seq_len, embed_dim)
        # Use the last token prediction (next token task)
        embed_last = embed[:, -1, :]  # (batch, embed_dim)
        logits = jnp.dot(embed_last, params['dense'])  # (batch, vocab_size)
        return logits

# Initialize model
vocab_size = 100  # Smaller vocab for easier learning
model = SimpleModel(vocab_size=vocab_size, embed_dim=32)
key = random.PRNGKey(0)
params = model.init_params(key)

def train_step(params, x, y, lr=0.1):
    def loss_fn(p):
        logits = model.forward(p, x)  # (batch, vocab_size)
        targets = y[:, -1]  # Predict next token
        # Stable cross-entropy
        loss = -jnp.mean(jnp.log(jax.nn.softmax(logits)[jnp.arange(logits.shape[0]), targets] + 1e-8))
        return loss

    loss, grads = jax.value_and_grad(loss_fn)(params)
    grads = jax.tree_util.tree_map(lambda g: jnp.clip(g, -1.0, 1.0), grads)
    params = jax.tree_util.tree_map(lambda p, g: p - lr * g, params, grads)
    return params, loss

# Warm-up
print("🔥 Warm-up JIT compilation...")
dummy_x = create_pattern_data(batch_size=8, seq_len=32)
dummy_y = create_pattern_data(batch_size=8, seq_len=32)
_ = train_step(params, dummy_x, dummy_y)
print("✅ JIT warm-up complete\n")

# Training loop
print("🚀 Starting training with PATTERN data...")
print("="*60)
print("Step | Loss | Time (s)")
print("-"*60)

step_times = []
losses = []

for step in range(50):
    x = create_pattern_data(batch_size=8, seq_len=32)
    y = create_pattern_data(batch_size=8, seq_len=32)

    start_time = time.perf_counter()
    params, loss = train_step(params, x, y)
    step_time = time.perf_counter() - start_time

    step_times.append(step_time)
    losses.append(float(loss))

    if step % 5 == 0:
        print(f"{step:4d} | {float(loss):8.4f} | {step_time:.4f}s")

print("="*60)

# Statistics
avg_time = np.mean(step_times)
total_time = np.sum(step_times)
loss_decrease = losses[0] - losses[-1]

print(f"\n✅ Training complete!")
print(f"📊 Average step time: {avg_time:.4f}s")
print(f"📊 Total training time: {total_time:.2f}s")
print(f"📊 Final loss: {losses[-1]:.4f}")
print(f"📊 Loss decrease: {loss_decrease:.4f} (from {losses[0]:.4f})")
print(f"💻 Device: {jax.devices()[0]}")

# Save metrics
device_name = str(jax.devices()[0]).replace(':', '_').replace(' ', '_')
with open(f"metrics_{device_name}.txt", "w") as f:
    f.write(f"Device: {jax.devices()[0]}\n")
    f.write(f"JAX Version: {jax.__version__}\n")
    f.write(f"Average step time: {avg_time:.4f}s\n")
    f.write(f"Total training time: {total_time:.2f}s\n")
    f.write(f"Initial loss: {losses[0]:.4f}\n")
    f.write(f"Final loss: {losses[-1]:.4f}\n")
    f.write(f"Loss decrease: {loss_decrease:.4f}\n")
    f.write("\nStep-by-step:\n")
    for i, (t, l) in enumerate(zip(step_times, losses)):
        f.write(f"{i},{t:.4f},{l:.4f}\n")

print(f"\n📁 Metrics saved to: metrics_{device_name}.txt")
```
